```python
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('C', 2), ('D', 5)],
    'C': [('A', 4), ('B', 2), ('D', 1)],
    'D': [('B', 5), ('C', 1), ('E', 3)],
    'E': [('D', 3)]
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 3,
    'D': 2,
    'E': 0
}

path, cost = a_star(graph, heuristic, 'A', 'E')

print("Shortest path:", path)
print("Total cost:", cost)
```

```python
import heapq


def a_star(graph, heuristic, start, goal):
    # Priority queue: (f_cost, current_node)
    open_set = [(heuristic[start], start)]

    # Cost from start to each node
    g_cost = {start: 0}

    # Store the previous node to reconstruct the path
    came_from = {}

    while open_set:
        f_cost, current = heapq.heappop(open_set)

        # Goal reached
        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]

            path.append(start)
            path.reverse()

            return path, g_cost[goal]

        # Explore neighbors
        for neighbor, edge_cost in graph[current]:
            new_g_cost = g_cost[current] + edge_cost

            # Found a cheaper path to neighbor
            if neighbor not in g_cost or new_g_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_g_cost
                came_from[neighbor] = current

                f_cost = new_g_cost + heuristic[neighbor]

                heapq.heappush(open_set, (f_cost, neighbor))

    # No path exists
    return None, float("inf")
```
